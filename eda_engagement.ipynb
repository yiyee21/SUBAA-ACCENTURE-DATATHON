```python
import pandas as pd
import matplotlib.pyplot as plt

attrition = pd.read_csv("attrition_log.csv")
employees = pd.read_csv("employees.csv")
engagement = pd.read_csv("engagement.csv")
performance = pd.read_csv("performance.csv")

def basic_eda(df):
    summary = pd.DataFrame({
        "dtype": df.dtypes,
        "missing": df.isna().sum(),
        "missing_%": df.isna().mean() * 100,
        "unique": df.nunique()
    })

    return summary.sort_values("missing_%", ascending=False)

basic_eda(attrition)
```

```python
engagement_vars = [
    "manager_effectiveness",
    "psychological_safety",
    "recognition",
    "career_development",
    "senior_leadership_trust",
    "purpose_meaning",
    "wellbeing",
    "confidence_in_role_future"
]
```

```python
engagement_stats = pd.DataFrame({
    "mean": engagement[engagement_vars].mean(),
    "median": engagement[engagement_vars].median(),
    "std": engagement[engagement_vars].std(),
    "min": engagement[engagement_vars].min(),
    "max": engagement[engagement_vars].max(),
    "skewness": engagement[engagement_vars].skew(),
    "missing": engagement[engagement_vars].isna().sum(),
    "missing_%": engagement[engagement_vars].isna().mean() * 100
})

display(engagement_stats.sort_values("mean", ascending=False))
```

```python
wave_means = engagement.groupby("wave_number")[engagement_vars].mean()

display(wave_means)
```

```python
engagement_driver_corr = (
    engagement[engagement_vars]
    .corr()["overall_engagement"]
    .drop("overall_engagement")
    .sort_values(ascending=False)
)

display(engagement_driver_corr)

plt.figure(figsize=(10, 6))

plt.barh(
    [x.replace("_", " ").title() 
     for x in engagement_driver_corr.index],
    engagement_driver_corr.values
)

plt.title("Correlation of Engagement Dimensions with Overall Engagement")
plt.xlabel("Correlation")
plt.ylabel("Engagement Dimension")

plt.tight_layout()
plt.show()
```

```python
engagement["engagement_level"] = pd.cut(
    engagement["overall_engagement"],
    bins=[0, 2, 3, 4, 5],
    labels=[
        "Very Low",
        "Low",
        "High",
        "Very High"
    ]
)

display(
    engagement["engagement_level"].value_counts()
)
```

```python
engagement_distribution = (
    engagement["engagement_level"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

display(engagement_distribution)

```
